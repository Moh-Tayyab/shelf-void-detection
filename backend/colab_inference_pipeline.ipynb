# 🛒 Shelf Void Detection — Inference Pipeline

Trained model (`best.pt`) se shelf images pe **void (khali)** aur **occupied (bhara)** detection.

| Class | Color | Matlab |

> Void/product class id model ke **asli naam se auto-detect** hote hain (`missing`, `shelf-void`, `void`, `empty` — koi bhi). Naam match na ho toh cell error dega, chupke galat fill% nahi dega.

## Step 2 — Model (`best.pt`) load karo

Trained model `best.pt` chahiye. **2 tareeke:**

**A. Auto (sharing ke liye best 👍):** `best.pt` apne Google Drive pe upload karo →
*Anyone with the link* share karo → link me se **file ID** copy kar ke neeche `BEST_PT_DRIVE_ID` me daalo.
(ID milega: `drive.google.com/file/d/<<<YEH_ID>>>/view`)

**B. Manual:** Drive ID khali chhodo, sidebar **📁 icon → `/content` → Upload → `best.pt`** select karo.

```python
# === MODEL LOAD ===
MODEL_PATH = "/content/best.pt"
BEST_PT_DRIVE_ID = ""   # <-- yahan apne Google Drive ka file ID paste karo (brackets nahi)

if BEST_PT_DRIVE_ID.strip():
    print("Drive se best.pt download ho raha hai (ID:", BEST_PT_DRIVE_ID.strip(), ") ...")
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "gdown"], check=False)
    import gdown
    gdown.download(id=BEST_PT_DRIVE_ID.strip(), output=MODEL_PATH, quiet=False)
    print()

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        "best.pt nahi mila!\n"
        "Ya to BEST_PT_DRIVE_ID me Drive file ID daalo,\n"
        "ya sidebar 📁 se /content/best.pt upload karo, phir ye cell dobara run karo."
    )

model = YOLO(MODEL_PATH)
class_names = model.names   # e.g. {0: 'shelf-void', 1: 'product'} (ya 3-class)

# --- Void (unoccupied) aur product (occupied) ke class id ROBUSTLY dhoondo ---
# Model ke asli class naam se match karte hain. Pehle yahan hardcoded
# `.get("missing", 0)` tha — model ka naam "shelf-void" tha, isliye lookup
# miss karke default 0 par gir gaya tha aur saare void galat se "product"
# count ho gaye the (missing_count=0, fill=100% ka bogus result). Ab fix.
_id_by_name = {v: k for k, v in class_names.items()}

def _find_class_id(*keywords):
    """Jis class ke naam me koi bhi keyword (case-insensitive) mile, uska id do — warna None."""
    for name, cid in _id_by_name.items():
        if any(kw in name.lower() for kw in keywords):
            return cid
    return None

MISSING_ID = _find_class_id("void", "missing", "empty", "gap")         # khali shelf
PRODUCT_ID = _find_class_id("product", "occupied", "item", "filled")   # bhara shelf

# Silent default HATANA — galat naam mile toh pakka error do, warna fill% galat niklega.
assert MISSING_ID is not None, (
    "❌ Void/missing class model.names me nahi mila: %r\n"
    "   data.yaml ka class naam check karke _find_class_id ke keywords badao." % class_names)
assert PRODUCT_ID is not None, (
    "❌ Product/occupied class model.names me nahi mila: %r" % class_names)

# Naam bhi rakh lo — area-based fill aur labels ke liye (aage use honge).
VOID_NAME = class_names[MISSING_ID]
PRODUCT_NAME = class_names[PRODUCT_ID]

print("\n✅ Model loaded :", MODEL_PATH)
print("   Classes     :", class_names)
print("   VOID  -> id %d (%r)  🔴 RED" % (MISSING_ID, VOID_NAME))
print("   PROD  -> id %d (%r)  🟢 GREEN" % (PRODUCT_ID, PRODUCT_NAME))
print("   Size        : %.1f MB" % (os.path.getsize(MODEL_PATH)/1024/1024))
```

## Step 4 — Settings (Confidence, IoU, NMS) + Color legend

- **CONFIDENCE** — kitna sure hona chahiye model ko (0.0 = sab dikhao, 1.0 = sirf guaranteed)
  - 0.10–0.20 → bohat zyada boxes, junk/false positives ke saath (pehle default 0.15 tha — overlap ka karan)
  - **0.22 (default)** → void recall ke liye thoda kam (void class minority hai); real detections rehte hain, junk drop
  - 0.30–0.50 → kam false positive chahiye toh (par void recall thoda kam)
- **IoU** — same-class overlapping boxes kitne similar hone par merge honge (NMS)
- **AGNOSTIC_NMS** — `True` (default): across-class NMS. Ek hi region pe product AUR void dono
  predict ho to higher-confidence wala rakha jaata hai. V6 me kuch regions dono class se label
  the (annotation conflict) — uski wajah se overlapping green+red boxes aate the. **Ye band-aid
  hai**, asli fix labels clean karke retrain karna hai (see `backend/clean_annotations.py`).
- **IMG_SIZE 960** — (640 se upar) thin void slivers survive karte hain.

Neeche wala cell **color-coded annotation helper** bhi define karta hai (🔴 missing / 🟢 product).

```python
# === TUNE THESE ===
CONFIDENCE = 0.22   # kam = zyada void detection (recall); zyada = kam false positive
                    # 0.30 -> 0.22: void class minority (~5% boxes) hai, recall ke liye thoda kam.
                    # (Pehle 0.15 bhi tha -> bohat junk aata tha. 0.22 balanced sweet-spot.)
IOU = 0.45          # NMS IoU threshold
AGNOSTIC_NMS = True # across-class NMS: ek region pe product AUR void dono predict ho to
                    # higher-confidence wala rakho. V6 annotations me kuch regions dono class
                    # se label the (conflict) -> model dono predict karta tha = overlapping
                    # green+red boxes. Ye band-aid hai; asli fix = labels clean karke retrain.
                    # (Band-aid ke saath bhi fill% thoda off ho sakta hai.)
IMG_SIZE = 960      # inference image size (640 -> 960): thin void slivers survive karte hain

print("Confidence   :", CONFIDENCE)
print("IoU          :", IOU)
print("Agnostic NMS :", AGNOSTIC_NMS)
print("Image size   :", IMG_SIZE)

# === COLOR-CODED ANNOTATION HELPER ===
# void/missing (unoccupied) -> RED, product (occupied) -> GREEN   (BGR for OpenCV)
COLOR_MISSING = (0, 0, 255)     # RED
COLOR_PRODUCT = (0, 255, 0)     # GREEN

def draw_and_count(results, src_path, out_path):
    """Color-coded boxes draw karke out_path pe save karta hai.
    Returns: (missing_n, product_n, detections_list).
    Har detection me 'area' (pixels²) bhi aata hai — area-based fill ke liye."""
    img = cv2.imread(src_path)
    if img is None:
        return 0, 0, []
    missing_n, product_n, dets = 0, 0, []
    for r in results:
        for box in r.boxes:
            cls = int(box.cls[0])
            conf = float(box.conf[0])
            x1, y1, x2, y2 = [int(v) for v in box.xyxy[0].tolist()]
            name = class_names[cls]
            area = max(0, x2 - x1) * max(0, y2 - y1)   # bbox area (area-based fill ke liye)
            color = COLOR_MISSING if cls == MISSING_ID else COLOR_PRODUCT

            cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
            text = name + " " + format(conf, ".2f")
            (tw, th), _ = cv2.getTextSize(text, cv2.FONT_HERSHEY_SIMPLEX, 0.5, 1)
            ytop = max(0, y1 - th - 6)
            cv2.rectangle(img, (x1, ytop), (x1 + tw + 4, ytop + th + 6), color, -1)
            cv2.putText(img, text, (x1 + 2, ytop + th + 2),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1, cv2.LINE_AA)
            if cls == MISSING_ID:
                missing_n += 1
            else:
                product_n += 1
            dets.append({"class": name, "confidence": round(conf, 4),
                         "bbox": [round(v, 1) for v in (x1, y1, x2, y2)],
                         "area": area})
    cv2.imwrite(out_path, img)
    return missing_n, product_n, dets

print("✅ Color-coded annotation ready: RED=%s (void), GREEN=%s (product)" % (VOID_NAME, PRODUCT_NAME))
```

## Step 5 — Single Image Inference (pehla test)

Pehli image pe test karo — boxes sahi aa rahe hain ya nahi.

```python
assert len(all_images) > 0, "Pehle /content/input me images upload karo (Step 3)."

test_img = all_images[0]
print("Testing:", os.path.basename(test_img), "\n")

results = model.predict(source=test_img, conf=CONFIDENCE, iou=IOU,
                        agnostic_nms=AGNOSTIC_NMS, imgsz=IMG_SIZE,
                        save=False, verbose=False)

os.makedirs("/content/single_test", exist_ok=True)
out_path = "/content/single_test/" + os.path.basename(test_img)
m, p, dets = draw_and_count(results, test_img, out_path)

print("Detections:", len(dets), " -> ", p, "product (green),", m, "missing (red)")
for d in dets[:12]:
    print("  %-8s conf=%.3f  bbox=%s" % (d["class"], d["confidence"], d["bbox"]))
if len(dets) > 12:
    print("  ... aur", len(dets)-12)

display(IPImage(filename=out_path, width=800))
```

## Step 6 — Batch Inference (sari images)

Ab saari images pe inference chalao. Har image ka color-coded version + CSV row banega.

```python
OUTPUT_DIR = "/content/inference_results"
ANNOTATED_DIR = OUTPUT_DIR + "/annotated"
os.makedirs(ANNOTATED_DIR, exist_ok=True)

csv_rows = []
print("Processing", len(all_images), "images...\n")

for i, img_path in enumerate(all_images):
    fname = os.path.basename(img_path)
    results = model.predict(source=img_path, conf=CONFIDENCE, iou=IOU,
                            agnostic_nms=AGNOSTIC_NMS, imgsz=IMG_SIZE,
                            save=False, verbose=False)
    out_path = os.path.join(ANNOTATED_DIR, fname)
    m, p, dets = draw_and_count(results, img_path, out_path)

    total = m + p

    # --- Area-based fill (headline metric) ---
    # Box-count fill (p/total) ek chhote aur ek bade void ko barabar maanta hai,
    # isliye "shelf fill %" ke liye AREA behtar hai:
    #   fill% = product area / (product area + void area)
    void_area = sum(d["area"] for d in dets if d["class"] == VOID_NAME)
    prod_area = sum(d["area"] for d in dets if d["class"] == PRODUCT_NAME)
    covered = void_area + prod_area
    fill_pct = round(100.0 * prod_area / covered, 1) if covered > 0 else None
    fill_by_count = round(100.0 * p / total, 1) if total > 0 else None

    csv_rows.append({
        "image": fname,
        "missing_count": m,
        "product_count": p,
        "total_detections": total,
        "shelf_fill_pct": fill_pct,           # area-based (asli metric)
        "fill_pct_by_count": fill_by_count,   # box-count based (reference)
        "void_area_px": void_area,
        "product_area_px": prod_area,
        "detections_json": json.dumps(dets),
    })

    flag = "⚠️ VOID" if m > 0 else "ok"
    print("  [%d/%d] %s: %d product, %d missing  -> fill %s%% (area) / %s%% (count)  %s" %
          (i+1, len(all_images), fname, p, m, fill_pct, fill_by_count, flag))

print("\n✅ Done -", len(all_images), "images processed ->", ANNOTATED_DIR)
```

## Step 7 — Results Summary Table

```python
df = pd.DataFrame(csv_rows)
df_display = df.drop(columns=["detections_json"])

print("=== DETECTION SUMMARY ===")
print("Total images            :", len(df))
print("Images with void/missing:", int((df["missing_count"] > 0).sum()))
print("Images fully occupied   :", int((df["missing_count"] == 0).sum()))
print("\nTotal missing (RED)   :", int(df["missing_count"].sum()))
print("Total product (GREEN) :", int(df["product_count"].sum()))
if len(df):
    print(f"Avg shelf fill %      : {df['shelf_fill_pct'].mean():.1f}%")
print()
display(df_display)
```

## Step 10 — Sample Results Gallery

Ek void wali + ek fully-occupied wali image dikhao.

```python
shown = set()
for row in csv_rows:
    category = "void" if row["missing_count"] > 0 else "occupied"
    if category in shown:
        continue
    path = os.path.join(ANNOTATED_DIR, row["image"])
    if not os.path.exists(path):
        continue
    if category == "void":
        print("\n--- VOID DETECTED:", row["image"], " (", row["missing_count"], "missing,", row["product_count"], "product) ---")
    else:
        print("\n--- FULLY OCCUPIED:", row["image"], " (0 missing,", row["product_count"], "product) ---")
    display(IPImage(filename=path, width=700))
    shown.add(category)
    if len(shown) >= 2:
        break
```
